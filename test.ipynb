```python
from llama_index.core import SimpleDirectoryReader
import os
path = "./File/Unstructured/Comparion of Diff AI Summary Results (test file-24Q1 BPDA Boston Economic Report)"

# filename_fn = lambda filename: {"file_name": filename}
filename_fn = lambda filename: {"file_name": os.path.basename(filename)}

# automatically sets the metadata of each document according to filename_fn
documents = SimpleDirectoryReader(
    path, file_metadata=filename_fn
).load_data()

print(len(documents))
print(documents[0].metadata)

# 打印文档数量
print(f"文档总数: {len(documents)}")

# 查看第一个文档的详细信息
print("\n第一个文档的信息:")
print(f"元数据: {documents[0].metadata}")
print(f"文档ID: {documents[0].doc_id}")
print(f"文本内容前500字符: \n{documents[0].text[:500]}...")

# 遍历所有文档的基本信息
print("\n所有文档的基本信息:")
for i, doc in enumerate(documents):
    print(f"\n文档 {i+1}:")
    print(f"id: {doc.doc_id}")
    print(f"文件名: {doc.metadata.get('file_name', 'N/A')}")
    print(f"文本长度: {len(doc.text)} 字符")
```
